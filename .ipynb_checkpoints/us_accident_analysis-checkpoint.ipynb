## Percentage of missing values per column

```python
missing_data_percentage=df.isna().sum().sort_values(ascending=False)/df.shape[0]*100
```

```python
missing_data_percentage = missing_data_percentage[missing_data_percentage>0]
missing_data_percentage
```

```python
plt.figure(figsize=(12,6))
ax = sns.pointplot(x=missing_data_percentage.index,
                   y=missing_data_percentage.values)
plt.xticks(rotation=90)
plt.xlabel('Column Names')
plt.ylabel('Missing Percentage')
plt.title('Percentage of missing data in the dataset')

for i, value in enumerate(missing_data_percentage.values):
    ax.text(i, value + 0.75,  
            f"{value:.2f}", 
            ha='center', va='bottom', fontsize=8, color='black')

plt.savefig('missing_percentage.png',format='png',dpi=300,bbox_inches='tight')
```

## Data Analysis and Visualization


```python
df.columns
```

```python
df['City'].unique()
```

```python
top_10_cities_accident=df['City'].value_counts()[:10]
top_10_cities_accident = pd.DataFrame(top_10_cities_accident)
top_10_cities_accident.reset_index(inplace=True)
top_10_cities_accident
```

```python
plt.barh(y=top_10_cities_accident['City'],width=top_10_cities_accident['count'])
plt.title('Top 10 cities with highest number of Accidents')
plt.xlabel('Accident Numbers')
plt.ylabel('City Names')
```
